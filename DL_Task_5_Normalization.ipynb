```python
np.set_printoptions(precision = 3, suppress = True)

x = np.array([ [1.0, 2.0, 30.0], 
              [2.0, 4.0, 32.0], 
              [3.0, 8.0, 34.0],
              [4.0, 8.0, 34.0]
             ]
             )

print(" Input Data : \n", x)
```

```python
eps = 1e-5

mean = x.mean(axis = 0) # mean of each column
var = x.var(axis = 0)  # variance of each column

print(" Mean Per Feature: ", mean)
print(" Variance Per Feature :", var)

x_hat = (x - mean)/ np.sqrt(var + eps)
print("\n Normalized x (x_hat) : \n", x_hat)
print("\n New Mean Per Feature (should be  -0): ", x_hat.mean(axis = 0))
print(" New Variance Per Feature (should be -1): ", x_hat.var(axis = 0))
```

```python
gamma = np.ones(3)
beta = np.zeros(3)

out = gamma* x_hat + beta
print(" Output After gamma/beta : \n", out)
```

```python
mean_in = x.mean(axis  = 1, keepdims = True)
var_in = x.var(axis = 1, keepdims = True)

x_hat_in = (x - mean_in)/ np.sqrt(var_in + eps)
print("LayerNorm Normalized x : \n", x_hat_in)
print("\n row means (should be -0) : ", x_hat_in.mean(axis = 1))
print(" row vars (should be -1) : ", x_hat_in.var(axis = 1))
                 
```
